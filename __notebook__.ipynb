# Initialize Selenium WebDriver

```python
def add_driver_options(options):
    """
    Add configurable options
    """
    chrome_options = Options()
    for opt in options:
        chrome_options.add_argument(opt)
    return chrome_options

def initialize_driver():
    """
    Initialize the web driver
    """
    driver_config = {
        "options": [
            "--headless",
            "--no-sandbox",
            "--start-fullscreen",
            "--allow-insecure-localhost",
            "--disable-dev-shm-usage",
            "user-agent=Chrome/116.0.5845.96"
        ],
    }
    driver = uc.Chrome(
        use_subprocess=False,
        headless=True,
    )
    return driver
```

# Helper Functions

```python
MAIN_URL = "https://www.mudah.my/malaysia/apartment-condominium-for-sale"

# get listings page
def get_listing_page_url_soup(page_num):
    url = f"{MAIN_URL}?o={page_num}"
    driver.get(url)
    soup = BeautifulSoup(driver.page_source, 'html')
    return soup

# get list of house url pages
def get_house_urls_for_page(soup):
    listings = soup.find('script', {'type': 'application/ld+json'})
    content = listings.contents[0]
    json_content = json.loads(content)
    item_list = next(entry for entry in json_content if entry["@type"] == "ItemList")["itemListElement"]
    house_urls = [item['item']['url'] for item in item_list]
    return house_urls

# get metadata for house url
def get_metadata_for_house_url(house_url):
    result = {}
    ref_ul_num = None
    
    driver.get(house_url)
    
    try:
        WebDriverWait(driver, 3).until(EC.presence_of_element_located((By.CLASS_NAME, 'border')))
        soup_house = BeautifulSoup(driver.page_source, 'html')
        
        # we first need to get the "reference" number for our classes, which we base off
        # the "ul" class which the tabs are stored in
        # i dunno, the housing pages are inconsistent
        uls = soup_house.find_all('ul')
        # to get the reference ul, we check if the first "li" in it has the text "Description"
        for ul in uls:
            lis = ul.find_all('li')
            if len(lis) == 2 and lis[0].text == "Description":
                ref_ul = ul
                ref_ul_num = int(ref_ul.attrs['class'][0][2:])
                break
        # default reference ul number if we cannot locate the ul
        if ref_ul_num is None:
            ref_ul_num = 214

        # first, get listing description
        description_text = soup_house.find('ul', {'class': f'mw{ref_ul_num+2}'}).get_text(separator="\n")
        result["description"] = description_text

        # then, switch tabs to "property details"
        property_details_tab = driver.find_elements(By.CLASS_NAME, f'mw{ref_ul_num+1}')[1]
        property_details_tab.click()
        soup_house = BeautifulSoup(driver.page_source, 'html')

        # we can then get all the relevant property metadata
        tables = soup_house.find('ul', {'class': f'mw{ref_ul_num+2}'}).findAll('table')
        for table in tables:
            table_rows = table.findAll('tr')
            for row in table_rows:
                entries = row.findAll('td')
                if len(entries) > 0:
                    key = entries[0].text
                    value = entries[1].text
                    result[key] = value
                    
        # finally, get the price
        price = soup_house.find('div', {'itemprop': 'offers'}).text
        result["price"] = price

        # return the result
        return result
    except Exception as e:
        return None
    
def get_df_from_results_list(results_list):
    results_df = pd.DataFrame.from_dict(results_list)
    return results_df
```

# Get The Data

```python
driver = initialize_driver()

results_list = []
pbar = tqdm(list(range(1, 101)))

for page_num in pbar:
    print(f"Getting data for page number {page_num}")
    soup = get_listing_page_url_soup(page_num)
    house_urls = get_house_urls_for_page(soup)
    for house_url in house_urls:
        pbar.set_postfix(page_num=page_num, house_url=house_url)
        result = get_metadata_for_house_url(house_url)
        if result is not None:
            results_list.append(result)
        else:
#             print(f"INFO: could not process {house_url}: trying again")
            # we got blocked by cloudflare
            # so wait a bit
            time.sleep(10)
            # and try again
            result = get_metadata_for_house_url(house_url)
            if result is not None:
                results_list.append(result)
                
    # every epoch, save our results to a CSV
    results_df = get_df_from_results_list(results_list)
    results_df.to_csv("houses.csv", index=False)
```
